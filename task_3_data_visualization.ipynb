```python
df = pd.read_csv("books_dataset (1).csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Price"], bins=10, edgecolor="black")

plt.title("Distribution of Book Prices")
plt.xlabel("Book Price (£)")
plt.ylabel("Number of Books")

plt.show()
```

```python
print("Insight: This histogram shows how book prices are distributed.")
print("It helps us identify the common price ranges and whether books are mostly cheap or expensive.")
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Rating", order=["One", "Two", "Three", "Four", "Five"])

plt.title("Number of Books by Rating")
plt.xlabel("Book Rating")
plt.ylabel("Number of Books")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Rating",
    y="Price",
    order=["One", "Two", "Three", "Four", "Five"],
    errorbar=None
)

plt.title("Average Book Price by Rating")
plt.xlabel("Book Rating")
plt.ylabel("Average Price (£)")

plt.show()
```

```python
avg_price = df.groupby("Rating")["Price"].mean()

print("Average book price by rating:")
print(avg_price)

print("\nInsight:")
print("This chart compares average book prices across different ratings.")
print("It helps us understand whether higher-rated books tend to cost more.")
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Price")

plt.title("Spread of Book Prices")
plt.xlabel("Book Price (£)")

plt.show()
```

Conclusion: These visualizations show the distribution of book prices, the number of books in each rating category, the average price by rating, and the spread of prices. They help us understand patterns in the dataset and communicate findings clearly.
